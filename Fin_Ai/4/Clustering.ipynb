<p dir=rtl style="direction: rtl; text-align: center; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    <b>توضیحات بالا و تابعی که در کلاس <code>Preprocessor</code> ایجاد کردیم، همگی مثال بودند و هیچ الزامی وجود ندارد که شما هم دقیقا همین تابع را در کلاس خود داشته باشید. پس با استفاده از خلاقیت و فکر، کلاسی توسعه دهید که به بهترین شکل داده را مهندسی می‌کند!</b>

```python
import numpy as np
import pandas as pd

class Preprocessor:
    def __init__(self, df):
        self.df = df.copy()
        
    def handle_missing_values(self):
        self.df.fillna(0, inplace=True)
        
    def drop_unnecessary_columns(self):
        # Drop columns not needed for training
        self.df.drop(columns=['ID_Order', 'ID_Customer', 'ID_Item'], inplace=True)
        
    def extract_time_features(self):
        # Convert 'DateTime_CartFinalize' to datetime
        self.df['DateTime_CartFinalize'] = pd.to_datetime(self.df['DateTime_CartFinalize'])
        
        # Extract time-based features
        self.df['hour'] = self.df['DateTime_CartFinalize'].dt.hour
        self.df['dayofweek'] = self.df['DateTime_CartFinalize'].dt.dayofweek
        self.df['month'] = self.df['DateTime_CartFinalize'].dt.month
        self.df['year'] = self.df['DateTime_CartFinalize'].dt.year
        
        # Drop original 'DateTime_CartFinalize'
        self.df.drop(columns=['DateTime_CartFinalize'], inplace=True)
        
    def encode_categorical_variables(self):
        # Encoding categorical variables (e.g., 'city_name_fa')
        self.df = pd.get_dummies(self.df, columns=['city_name_fa'], drop_first=True)
        
    def transform(self):
        # Apply all preprocessing steps
        self.handle_missing_values()
        self.drop_unnecessary_columns()
        self.extract_time_features()
        self.encode_categorical_variables()
        
        return self.df
```

انتخاب تعداد خوشه‌ها و آموزش مدل
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    حال که داده را پاکسازی کرده و احتمالا ویژگی‌هایی را به آن افزوده یا از آن حذف کرده‌اید، وقت آن است که ابتدا براساس روش‌هایی که می‌دانید تعداد خوشه‌های مناسب برای این سوال را پیدا کنید و سپس بر اساس آن مدلی آموزش دهید که بتواند دسته‌بندی هر مشتری را مشخص کند. <br>راهنمایی: برای انتخاب تعداد خوشه‌ها اگر از معبار <code>silhouette score</code> استفاده کنید، برای هر <code>k</code> حداقل به ۳ الی ۴ دقیقه زمان نیاز خواهید داشت!
</font>
</p>

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

class ClusterModel:
    def __init__(self, df):
        self.df = df
        self.model = None

    def find_optimal_k(self):
        # List to store silhouette scores for different k values
        silhouette_scores = []

        # Testing k values from 2 to 10
        for k in range(2, 11):
            kmeans = KMeans(n_clusters=k, random_state=42)
            kmeans.fit(self.df)
            score = silhouette_score(self.df, kmeans.labels_)
            silhouette_scores.append(score)
        
        # Plot silhouette scores to visualize the best k
        plt.plot(range(2, 11), silhouette_scores)
        plt.xlabel('Number of Clusters (k)')
        plt.ylabel('Silhouette Score')
        plt.title('Silhouette Score for Different k Values')
        plt.show()

        # Return the best k (with the highest silhouette score)
        optimal_k = silhouette_scores.index(max(silhouette_scores)) + 2
        return optimal_k

    def train_model(self, k):
        # Train KMeans model with the optimal k
        self.model = KMeans(n_clusters=k, random_state=42)
        self.model.fit(self.df)

    def predict(self):
        # Predict clusters for the data
        return self.model.predict(self.df)
```

```python
# Assuming you already have the cleaned and preprocessed data
# Preprocess the data using the Preprocessor class
preprocessor = Preprocessor(df)
df_processed = preprocessor.transform()

# Now, let's initialize and use the ClusterModel class
cluster_model = ClusterModel(df_processed)

# Find the optimal number of clusters (k) based on silhouette score
optimal_k = cluster_model.find_optimal_k()

# Train the KMeans model with the optimal k
cluster_model.train_model(optimal_k)

# Predict clusters for the data
cluster_labels = cluster_model.predict()

# If you want to add the cluster labels back to the original DataFrame
df_processed['Cluster_Label'] = cluster_labels
```
